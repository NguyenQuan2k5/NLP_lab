## Thử nghiệm dự đoán với tin nhắn mới

Sau khi cell huấn luyện ở trên hoàn thành, chạy cell bên dưới để kiểm tra khả năng dự đoán. Bạn có thể thay đổi các câu trong danh sách `examples` bằng tin nhắn của riêng mình.

```python
import joblib
model = joblib.load("intent_model.joblib")
examples = [
    "I want my money back, please.",
    "Please cancel my order.",
    "Can you tell me where my package is?",
    "My payment was declined."
]
for message, intent in zip(examples, model.predict(examples)):
    print(f"Tin nhắn: \"{message}\" -> Dự đoán ý định: [{intent}]")
```
